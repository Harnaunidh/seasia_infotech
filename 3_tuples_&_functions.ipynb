```python
# Write a Python function that takes a list of integers and returns a new list with the 
# numbers sorted first by their odd/even nature (odd numbers first), and then by the 
# value (in ascending order for both odd and even numbers).

list1 = []
list2 = []
list3 = []
n = int(input("enter the no. of elements: "))

for i in range(n):
    num = int(input("enter numbers: "))
    list1.append(num)
for j in list1:
    if j%2 != 0:
        list2.append(j)
for k in list1:
    if k%2 == 0:
        list3.append(k)
print (list2 + list3)
list2.sort()
list3.sort()
print (list2 + list3)
```

```python
# write a function that counts the number of vowels in a given string.

def vowel_count(word):
    word = input("enter a word: ")
    vowels = ['a', 'e', 'i', 'o', 'u']
    count = 0

    for i in word:
        if i in vowels:
            count +=1
    print(count)
vowel_count(word)
```

```python
#  Write a Python function that accepts two lists of strings.
#  Return a list of all unique words that appear in both
#  lists (ignoring case), and return them in alphabetical order.

list1 = []
list2 = []
list3 = []

for i in range(5):
    word = input("enter the words for list 1 ")
    list1.append(word)

for i in range(5):
    word = input("enter the words for list 2 ")
    list2.append(word)

list1.extend(list2)

for j in list1:
    if (j not in list3):
        list3.append(j)

list3.sort()
print(list3)
```
