```python
# Write a function that takes a number as input and returns the square root of the number. 
# If the input is negative, return a message indicating the input is invalid.

import math

num = int(input("enter a number: "))
if num >= 0:
    print(f"Squareroot of {num} is {math.sqrt(num)}")

else:
    print("invalid input")
```

```python
# Write a function that calculates the area of a circle given its radius, using the formula: Area = π * radius².

import math

radius = int(input("enter the radius of the circle: "))
if radius >= 0:
    area = math.pi*radius**2
    print(area)

else:
    print("invalid input")
```

```python
# Write a Python function that generates a random password consisting of 
# uppercase letters, lowercase letters, and digits (8 characters long).

import random
import string

char = string.ascii_letters + string.digits
password = ''.join(random.choices(char, k=8))
print(f"Password: {password}")

```

```python
# Write a Python function that calculates the number of days between two dates.

from datetime import datetime

def date_dif(a ,b , date_format='%Y-%m-%d'):
    a1 = datetime.strptime(a , date_format)
    b1 = datetime.strptime(b , date_format)
    return (b1 -a1).days

c = '2003-12-01'
d = '2004-08-02'
print(f"The difference of days between {c} and {d} is {date_dif(c, d)} days")
```
