```python
import sqlite3

# Establish connection to database (or create one if it doesn't exist)
conn = sqlite3.connect('myDatabase.db')

# Create a cursor object using the connection
cursor = conn.cursor()
```

```python
cursor.execute('''
CREATE TABLE IF NOT EXISTS users (
    id INTEGER PRIMARY KEY,
    name TEXT,
    age INTEGER
);
''')

# Commit the changes to save them to the database
conn.commit()
```

```python
cursor.execute('''
INSERT INTO users (name,age) VALUES (?,?)
''', ('John Doe', 30))

conn.commit()
```

```python
cursor.executemany('''
INSERT INTO users (name, age) VALUES (?, ?)
''', [('Alice', 28), ('Bob', 35)])

conn.commit()
```

```python
cursor.execute('SELECT * FROM users')
rows = cursor.fetchall()

for row in rows:
    print(row)
```

```python
cursor.execute('''
UPDATE users SET age = 31 WHERE name = 'John Doe'
''')

conn.commit()
```

```python
cursor.execute("DELETE FROM users WHERE id=2")
conn.commit()
```

```python
cursor.executemany('''
INSERT INTO users (name, age) VALUES (?, ?)
''', [('John', 29), ('Rick', 35)])

conn.commit()
```

```python
cursor.execute("drop table users")
conn.commit()
```
