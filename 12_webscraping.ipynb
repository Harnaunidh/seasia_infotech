```python
product_name1 = []
product_price1 = []
product_rating1 = []

url = "https://www.flipkart.com/search?p%5B%5D=facets.brand%255B%255D%3DSamsung&sid=tyy%2F4io&sort=recency_desc&ctx=eyJjYXJkQ29udGV4dCI6eyJhdHRyaWJ1dGVzIjp7InRpdGxlIjp7Im11bHRpVmFsdWVkQXR0cmlidXRlIjp7ImtleSI6InRpdGxlIiwiaW5mZXJlbmNlVHlwZSI6IlRJVExFIiwidmFsdWVzIjpbIkxhdGVzdCBTYW1zdW5nIG1vYmlsZXMgIl0sInZhbHVlVHlwZSI6Ik1VTFRJX1ZBTFVFRCJ9fX19fQ%3D%3D&wid=1.productCard.PMU_V2_1&page=2"

headers = {
        "User-Agent": "Mozilla/5.0",
        "Accept": "application/json",
    }
web3 = requests.get(url, headers=headers)
soup = BeautifulSoup(web3.content, "lxml")

box = soup.find("div", class_ = "DOjaWF gdgoEp")

name = box.find_all("div", class_="KzDlHZ")
for n in name:
    product_name1.append(n.text)

price = box.find_all("div", class_="Nx9bqj _4b5DiR")
for n in price:
    product_price1.append(n.text)

rating = box.find_all("div", class_="XQDdHH")
for n in rating:
    product_rating1.append(n.text)
```

```python
data1 = {"name": product_name1,
        "price": product_price1,
        "rating": product_rating1}
pd.DataFrame(data1)
```

```python

product_name = []
product_price = []
product_rating = []

for i in range(1,10):

    url = "https://www.flipkart.com/search?p%5B%5D=facets.brand%255B%255D%3DSamsung&sid=tyy%2F4io&sort=recency_desc&ctx=eyJjYXJkQ29udGV4dCI6eyJhdHRyaWJ1dGVzIjp7InRpdGxlIjp7Im11bHRpVmFsdWVkQXR0cmlidXRlIjp7ImtleSI6InRpdGxlIiwiaW5mZXJlbmNlVHlwZSI6IlRJVExFIiwidmFsdWVzIjpbIkxhdGVzdCBTYW1zdW5nIG1vYmlsZXMgIl0sInZhbHVlVHlwZSI6Ik1VTFRJX1ZBTFVFRCJ9fX19fQ%3D%3D&wid=1.productCard.PMU_V2_1&page="+str(i)

    headers = {
            "User-Agent": "Mozilla/5.0",
            "Accept": "application/json",
        }

    web3 = requests.get(url, headers=headers)
    soup = BeautifulSoup(web3.content, "lxml")

    box = soup.find("div", class_ = "DOjaWF gdgoEp")
    name = box.find_all("div", class_="KzDlHZ")
    for n in name:
        product_name.append(n.text)
    
    price = box.find_all("div", class_="Nx9bqj _4b5DiR")
    for n in price:
        product_price.append(n.text)

    rating = box.find_all("div", class_="XQDdHH")
    for n in rating:
        product_rating.append(n.text)

print(len(product_name))
print(len(product_price))   
print(len(product_rating))           
```
