```python
# Create a class called Person with the following attributes: name, age, and gender.
# Define a method introduce() that prints a message introducing the person, including
# their name, age, and gender.

class Person:
    def __init__(self, name, age, gender):
        self.name = name
        self.age = age
        self.gender = gender

    def introduce(self):
        print(f"Hi my name is {self.name}. I am a {self.age} years old {self.gender}")
        
a1 = Person("Bobby", 18, "Male")
a1.introduce()

```

```python
# Create a class called Rectangle that has two attributes: length and width. 
# Implement a method area() to calculate the area of the rectangle. Also, 
# implement a method perimeter() to calculate the perimeter of the rectangle.

class Rectangle:
    def __init__(self, length, width):
        self.length = length
        self.width = width

    def area(self):
        return self.length*self.width
    
    def perimeter(self):
        return 2*self.length + 2*self.width
    
a1 = Rectangle(12,16)
print(a1.area())
print(a1.perimeter())
    
        



```

```python
# Create a class called BankAccount with attributes account_holder and balance. Implement methods to:

#     deposit(amount): Increases the balance by the given amount.
#     withdraw(amount): Decreases the balance by the given amount.
#     check_balance(): Returns the current balance.

# Also, handle cases where the withdrawal amount is greater than the current balance (insufficient funds).
        
class BankAccount:

    def __init__(self, account_holder, balance):
        self.account_holder = account_holder
        self.balance = balance
        print(f"current balance: {self.balance}"  )

    def deposit(self, amount):
        self.amount = amount
        # print(self.balance+amount)
        self.balance+=amount
        return self.balance 
        
    
    def withdraw(self, amount):
        self.amount = amount
        self.balance-=amount
        return self.balance
    
    def check_balance(self):
        return self.balance
    
a1 = BankAccount("Jon", 10000)
a1.deposit(6000)
a1.withdraw(2000)        



```

```python
# Create a class Book with the following attributes: title, author, price, and year_published. 
# Implement a method discount_price(discount_percentage) that returns the price of the book 
# after applying the discount.

class Book:

    def __init__(self, title, author, price, year):
        self.title = title
        self.author = author
        self.price = price
        self.year = year

    def disc_price(self, percentage):
        self.percentage = percentage
        return self.price - (self.price*(self.percentage/100))
    
a1 = Book("good", "me" , 400, 2004)
a1.disc_price(20)

```

```python
# Create a class Student that has the following attributes: name, roll_number, and marks. 
# Write a method grade() that assigns a grade based on the marks:

# 90 and above: "A"
# 70-89: "B"
# 50-69: "C"
# Below 50: "D"

class Student:

    def __init__(self, name, roll_no, marks ):
        self.name = name
        self.roll_no = roll_no
        self.marks = marks

    def grade(self):
        if self.marks >= 90:
            print("Grade: A")
        elif 89>=self.marks>=70:
            print("Grade: B") 
        elif 69>=self.marks>=50:
            print("Grade: C") 
        elif 50>=self.marks:
            print("Grade: D") 

a1 = Student("jon", 21, 62)
a1.grade()        




```

```python
# Create a class called Shape with a method area(). Then create two subclasses: Circle and Square. 
# Each subclass should have its own area() method:

# Circle should accept a radius and calculate the area as π * r^2.
# Square should accept a side and calculate the area as side^2.

class Shape:

    def area(self):
        pass

class circle(Shape):
    def __init__(self, radius):
        self.radius = radius

    def area(self):
        return 3.14*self.radius**2
    
class square(Shape):
    def __init__(self, side):
        self.side = side

    def area(self):
        return self.side**2
    
a1 = square(6)
a1.area()
        
```

```python
# Create a class Employee with the following attributes: name, id, department, and salary. Implement methods:

#     display_details(): Displays the employee’s details.
#     increase_salary(percent): Increases the salary by a given percentage.

# Create a subclass Manager that also includes an attribute for team_size. Add a method to display the manager’s team size.

class Employee:

    def __init__(self, name, id, dept, salary):
        self.name = name
        self.id = id
        self.dept = dept
        self.salary = salary

    def display_details(self):
        return (f'''Name: {self.name}
Employee id: {self.id}
Department: {self.dept}
Salary: {self.salary}''')
    
    def sal_inc(self, percent):
        self.percent = percent
        return (f"Salary will icrease to: { self.salary + (self.salary*(self.percent/100))}")
    
class Manager:

    def __init__(self, name, id, dept, salary, team_size):
        super().__init__(name, id, dept, salary)
        self.team_size = team_size

    def display_details(self):
        print(f'''Name: {self.name}
Employee id: {self.id}
Department: {self.dept}
Salary: {self.salary}
Team Size: {self.team_size}''')
        
a1 = Employee("John", 312, "IT", 35000)
print(a1.display_details())
print(a1.sal_inc(15))


```

```python
# Create a class ElectricCar that inherits from a Car class. Add an attribute battery_size 
# and a method charge() that prints a message saying that the car is charging. 
# Override the display_details() method to also include the battery size.

class Car:

    def __init__(self, make, model, year, color):
        self.make = make
        self.model = model
        self.year = year
        self.color = color

    def display_details(self):
        return f"{self.year} {self.color} {self.make} {self.model}"
    
class ElectricCar(Car):

    def __init__(self, make, model, year, color, battery_size):
        super().__init__(make, model, year, color)
        self.battery_size = battery_size

    def charge(self):
        print(f"The {self.make} {self.model} is charging...")

    def display_details(self):
        car_details = super().display_details() 
        return f"{car_details} with a {self.battery_size}-kWh battery"
    
ec1 = ElectricCar("Tesla", "Model S", 2023, "red", 1000)
print(ec1.display_details())
ec1.charge()
        
```
