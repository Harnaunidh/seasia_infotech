```python
# Create a Python program that:

# Prompts the user to input an integer.
# If the user enters a non-integer value, catch the ValueError and print an appropriate error message.
# If the user enters a number less than 0, catch the NegativeNumberError (custom error) and print an appropriate message.

class NegativeNumberError(Exception):
    def __init__(self, message="Number must be non-negative"):
        self.message = message
        super().__init__(self.message)

def get_positive_integer():
    try:
        user_input = int(input("Please enter a positive integer: "))
        
        if user_input < 0:
            raise NegativeNumberError
        
        print("Valid input")
    
    except ValueError:
        print("Error: Please enter a valid integer.")
    
    except NegativeNumberError as e:
        print(f"Error: {e}")
    
    except Exception as e:
        print(f"An unexpected error occurred: {e}")

get_positive_integer()

```

```python
# Write a Python program that:

# Prompts the user to input an age.
# If the input is not an integer, raise a ValueError.
# If the input is less than 18, raise a custom exception called AgeNotEligibleException.

class AgeNotEligibleException(Exception):

    def __init__(self, msg="You are not eligible"):
        self.msg = msg
        super().__init__(self.msg)

    def CheckAge():

        try:
            age = int(input("enter your age: "))
            if age>=18:
                print("You are eligible")
            elif age<0:
                raise NegativeNumberError
            else:
                raise AgeNotEligibleException
            
        except AgeNotEligibleException as e:
            print(e)

        except ValueError:
            print("Error: Please enter a valid integer.")

        except NegativeNumberError as e:
            print(f"Error: {e}")
            
    CheckAge()

```

```python
# Write a Python function that calculates the square root of a number. 
# The function should raise an AssertionError if the input is a negative number, 
# asserting that the number should be positive for a square root calculation.

import math

try:
    num = int(input("enter the number: "))
    assert num>=0
    print(f"Squreroot of {num} is {math.sqrt(num)}")   
    
except AssertionError:
    print("Error: enter a positive number.")

```

```python
# Write a Python program that tries to access an element in a list at an index provided by the user. 
# If the index is out of range, handle the IndexError and print a message explaining the error.

list = [12,16,24,21,56,39,42]
try:
    index= int(input("enter the index: "))
    
    if (index<0):
        raise NegativeNumberError
    elif (index>=len(list)):
        raise IndexError
    else:
        print(list[index])

except ValueError:
    print("Error: Please enter a valid integer.")

except NegativeNumberError as e:
    print(f"Error: {e}")

except IndexError:
    print(f"Index {index} is not in the list.")


```

```python
# Write a Python program that performs division of two numbers. Ensure that:

# If a ZeroDivisionError is raised, print a message saying, "Cannot divide by zero."
# Regardless of whether an exception occurs or not, always print the message "Operation Complete" using a finally block.

try:
    n1 = int(input("enter first number: "))
    n2 = int(input("enter second number: "))

    print(f"{n1} divided by {n2} is {n1/n2}")

except ValueError:
    print("Error: Please enter a valid number.")

except ZeroDivisionError:
    print("Error: you cannot divide a number by zero")

finally:
    print("Operation Complete.")
```
