```python
#Create a filter array that will return only values higher than 30:
arr = np.array((23,34,45,56,67))
filter_arr= []

for element in arr:
  if element  > 30:
    filter_arr.append(True)
  else:
    filter_arr.append(False)

newarr = arr[filter_arr]
print(filter_arr)
print(newarr)
```
