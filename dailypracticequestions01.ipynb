```python
salary = int(input("enter your income: "))

if (salary <= 10000):
    print("rate of interest: 0")
elif(10000<salary<=20000):
    print(f"rate of interest: {(salary-10000) * 1/10}")
else:
    print(f"rate of interest: {1000 + (salary-20000) * 2/10}")
```

```python
#find all the prime numbers between two given numbers.

a = int(input("enetr the first number :"))
b = int(input("enetr the second number :"))

for i in range (a,b+1):
    count=0
    for k in range(1,i+1):
        if i%k == 0:
            count+=1
    if count < 3:
        print(i)
```
