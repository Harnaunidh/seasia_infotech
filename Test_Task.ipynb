```python
# Ans 1.

name= input("enter the company's name: ")
char = []

for i in set(name):
    count=0
    for j in name:
        if (i==j):
            count+=1
    char.append((i, count))

def count(item):
    return item[1]

char.sort(key=count, reverse=True)

for chr, cnt in char[:3]:
    print(chr, cnt)
```

```python
# Ans 2.

def kth_row(arr):
    return arr[k]


athletes = input("enter the no. of athletes and no. of attributes seperated by space: ")

att = athletes.split(" ")
n = int(att[0])

arr = []

for i in range(n):
    atr = input("enter the attributes: ")
    arr.append(atr.strip().split())

k = int(input())

arr.sort(key=kth_row)

for row in arr:
    print(*row)
```

```python
# Ans 3.

from datetime import datetime

time_format = "%a %d %b %Y %H:%M:%S %z"

test_cases = int(input())

for i in range(test_cases):

    t1= input("enter time in the format-'Day dd Mon yyyy hh:mm:ss +xxxx': ").strip()
    t2= input("enter time in the format-'Day dd Mon yyyy hh:mm:ss +xxxx': ").strip()

    dt1= datetime.strptime(t1, time_format)
    dt2= datetime.strptime(t2, time_format)

    diff= abs(int((dt1 - dt2).total_seconds()))

print(diff)
```
